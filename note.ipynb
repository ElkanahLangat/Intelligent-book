```python
# A simple chatbot that asks for and remembers your name


def simple_chatbot():
    print("Bot: Hello! I am your AI assistant.")

    # Ask the user for their name and save it to a variable
    user_name = input("Bot: What is your name? -> ")

    print(f"Bot: It is great to meet you, {user_name}!")

    # Ask the bot if it remembers
    question = input(
        f"\n{user_name}: Do you know my name? (Type 'yes' or 'no') -> "
    )

    if "yes" in question.lower():
        print(f"Bot: Of course! Your name is {user_name}.")
    else:
        print(f"Bot: Well, you just told me your name is {user_name}!")


# Run the bot
simple_chatbot()
```

```python
# The Interactive Knowledge-Base Chatbot

# 1. PASTE YOUR DATASET HERE
# Replace this text with whatever documents, notes, or facts you want!
MY_DATASET = """
Project Orion is a secret space exploration mission launching in November 2027.
The mission commander is Captain Alex Mercer.
The target destination is Europa, a moon of Jupiter, looking for water ice.
The trip will take exactly 3 years using advanced ion propulsion engines.
"""


def smart_qa_bot():
    print("Bot: Hello! I am scanning your dataset right now...")
    print("Bot: Done! I am ready to answer your questions based on the text.")
    print(
        "----------------------------------------------------------------------"
    )

    # 2. RUN A LOOP TO KEEP THE CONVERSATION GOING
    while True:
        # Get the user's question
        user_question = input("\nYou: ")

        # Create a way to exit the loop
        if user_question.lower() in ["exit", "quit", "stop", "bye"]:
            print("Bot: Goodbye! Have a great day.")
            break

        # 3. CORE AI SEARCH FEATURE
        # We split the dataset into individual sentences
        sentences = MY_DATASET.strip().split("\n")
        found_answer = False

        # Scan each line of the dataset to look for matching keywords
        for sentence in sentences:
            # Clean up the sentence and question for matching
            words = user_question.lower().replace("?", "").split()

            # Check if the sentence contains the keywords you asked about
            if any(word in sentence.lower() for word in words if len(word) > 3):
                print(f"Bot: According to your dataset: '{sentence}'")
                found_answer = True
                break

        if not found_answer:
            print(
                "Bot: I couldn't find that specific information in your dataset. Try asking about 'commander', 'destination', or 'launch date'."
            )


# Start the application
smart_qa_bot()
```
