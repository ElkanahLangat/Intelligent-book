```python
# A simple chatbot that asks for and remembers your name


def simple_chatbot():
    print("Bot: Hello! I am your AI assistant.")

    # Ask the user for their name and save it to a variable
    user_name = input("Bot: What is your name? -> ")

    print(f"Bot: It is great to meet you, {user_name}!")

    # Ask the bot if it remembers
    question = input(
        f"\n{user_name}: Do you know my name? (Type 'yes' or 'no') -> "
    )

    if "yes" in question.lower():
        print(f"Bot: Of course! Your name is {user_name}.")
    else:
        print(f"Bot: Well, you just told me your name is {user_name}!")


# Run the bot
simple_chatbot()
```
